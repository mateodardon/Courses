```python
import pandas as pd 

#Cree un DataFrame para almacenar datos meteorológicos
metereo = {'fecha' : ['12-jun-26', '13-jun-26','14-jun-26', '15-jun-26', '16-jun-26'], 
           'Grados C': [20,19,17,21,20], '%% humedad': [65, 52, 54, 84, 50],
            'condicion' : ['Soleado', 'nublado', 'lluvioso', 'soleado', 'nublado'] }
df = pd.DataFrame(metereo)

#Establezca el índice en otra columna como fechas
a = df.set_index('fecha')
print(a)
```

```python
#Seleccionar una fila desde el punto medio utilizando iloc
medio = len(df)//2
df.iloc[medio]
```

```python
#Filtre las filas que coincidan con determinados criterios numéricos con loc
grades_mask = df.loc[:,'Grados C'] <20
df[grades_mask]
```

```python
#otra version del codigo anterior 
df.loc[df['Grados C'] > 19]
```

```python
#Filtrar un DataFrame para mostrar sólo los valores superiores a la media
avr_grades = df['Grados C'].mean()
df.loc[df['Grados C'] > avr_grades]
```

```python
#Combinar filtros para encontrar filas que coincidan con varias condiciones booleanas

df.loc[(df['Grados C'] > avr_grades) & (df['condicion'] == 'nublado')]
```

```python
#Utilice isin() para filtrar datos categóricos como los tipos de producto

df.loc[df['condicion'].isin(['lluvioso', 'nublado'])]
```

```python
#Rebane DataFrame con .loc[] utilizando un filtro booleano

df.loc[(df['Grados C']>19), ['fecha','condicion']]
```

```python
#Encadene filtros complejos con múltiples operadores booleanos

df.loc[(df['Grados C']>19) & ((df['condicion'] == 'lluvioso')| (df['condicion'] == 'soleado'))]
```
