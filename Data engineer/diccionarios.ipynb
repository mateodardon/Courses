```python
#¿Cuándo querría utilizar un diccionario frente a una lista o un conjunto?

diccionario = {'Mateo':"5548427622", 'Diego':"354848431", 'Margarita': "684344549"}

diccionario['Salvador'] = '547941383'
for nombre, telefono in diccionario.items():
    print(nombre, ": " ,telefono)

#Construya un conjunto de palabras únicas a partir de una cadena de entrada larga

texto = "El perro corretea el gato perro"

cadena = texto.split()
a = set(cadena)
print(a)

#Compare matrices de números utilizando intersecciones y diferencias de conjuntos
grupo_a = {1, 2, 3, 4, 5}
grupo_b = {4, 5, 6, 7, 8}

print(grupo_a.intersection(grupo_b))
print(grupo_a.difference(grupo_b))

#Ordene un diccionario por clave y por valor
edades = {"Carlos": 30, "Ana": 25, "Beto": 40}

orden = sorted(edades.items(), key = lambda par : par[1], reverse=True)
print(orden)
#Cree histogramas de recuentos de palabras utilizando diccionarios
histo= {}
for palabra in cadena:
    if palabra in histo:
        histo[palabra] += 1
    else:
        histo[palabra] = 1
print (histo)
```

```python
# Generator function yields numbers one by one 
def num_sequence(n):
    for i in range(n): 
        yield i

seq = num_sequence(5)
print(next(seq)) # 0
print(next(seq)) # 1 
```

```python
text = "ser o no ser"
conteo = {}
for palabra in text.split():
    conteo[palabra] = conteo.get(palabra, 0) + 1

print(conteo)
```
